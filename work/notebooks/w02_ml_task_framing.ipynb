## 5. Why ML beats a fixed rule here

A fixed rule like "stale AND visible" only checks two conditions with
hard-coded thresholds (180 days, 500 impressions) — it can't discover
non-obvious combinations or the right thresholds on its own. In Week 1-2
testing, that hand rule reached Precision@50 = 0.240, while a learned
model reached up to 0.740 — a ~3x improvement. This gap suggests the
real signal in the data is more complex than a two-condition rule can
capture, which is exactly where ML earns its place: finding patterns a
human wouldn't think to hard-code.
